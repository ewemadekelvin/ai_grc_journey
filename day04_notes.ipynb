# Course 3 Notes – Discover the Art of Prompting ✅ (5 Oct 2026)

## 1. What is a prompt?
The input given to an AI to get a desired output.
The model uses **what it learned in training** + **what I give it in the prompt**
(instructions, context, documents, examples).

## 2. LLM – Large Language Model
Trained on huge amounts of text; generates answers by **predicting the most
likely next word**. Explains hallucination: it produces what *sounds* likely,
not always what is true.

## 3. Prompting framework: T-C-R-E-I
| Step | Meaning | Example |
|---|---|---|
| **Task** | What to do — start with a **verb**; add format/persona | "Summarise this report in 5 bullets as a risk analyst" |
| **Context** | Background, audience, purpose | "for a non-technical manager" |
| **References** | Examples or guidelines | "follow this sample summary" |
| **Evaluate** | Human checks the output | Accurate? Fits the purpose? |
| **Iterate** | Refine and retry | Rephrase, add detail, split the task |

## 4. Prompting techniques
- **Chain-of-thought:** ask the AI to reason **step by step** before answering.
- **Prompt chaining:** break a big task into steps; **output of one prompt = input to the next**
  (e.g. outline, then draft, then edit).
- Can combine both.

## 5. Limitation – the context window
- **Context window** = how much text a model can hold in mind at once.
- Very long chats → earlier details get lost.
- **Fixes:** break work into shorter chats, use bullet checkpoints, or
  carry a **Markdown summary** into a new chat (what I do in practice).
