## 1. Harm
**My definition:** the use of AI in any context that may deprive certain groups
or organisations of access to information, resources or opportunities, or damage
their reputation in a way that is hurtful.

| Type | My explanation | Example |
|---|---|---|
| **Allocative** | AI withholds information, resources or opportunities based on race or social values learned from its training data | Loan or job applications unfairly rejected for one group |
| **Quality-of-service** | AI does not perform the same way for all users | Facial recognition far less accurate for darker-skinned women (*Gender Shades* study) |
| **Representational** | AI reinforces harmful stereotypes about a group | Google Photos (2015) labelled Black people as "gorillas", caused by unrepresentative training data and poor testing |
| **Social system** | Wider harm to society | Deepfakes spreading misinformation |
| **Interpersonal** | AI used to harm an individual | Fake images used for harassment |

**GRC lesson:** Google's fix was to remove the label — patching the symptom, not the cause.

## 2. Bias
- **Where it comes from:** skewed or unrepresentative training data, which produces
  inaccurate or unfair output. **Systemic bias** in society also flows into the data.
- **How to reduce it:** use more representative data, test results for each group
  separately, involve diverse reviewers, and keep a human in the loop.

## 3. Three terms not to mix up
| Term | What it is | Cause |
|---|---|---|
| **Context window** | Model loses track of details in a long chat | Limit on how much text it can hold at once |
| **Knowledge cutoff** | Model doesn't know events after its training date | Training data stops at a fixed date |
| **Drift** | A deployed model's accuracy declines over time | The real world changes while the model stays the same |

## 4. Privacy & security
**My rule:** protect sensitive information by **never entering real personal details**
into AI tools — use **placeholders** instead, give a specific prompt to guide the
output, then **verify** the result with a human-in-the-loop approach.
- GRC terms: **data minimisation / pseudonymisation**.
- **Why it matters:** data entered may be stored, reviewed or used for training;
  sharing personal data without a lawful basis can breach **UK GDPR** —
  creating legal, financial and reputational risk.

## 5. My quant edge – fairness check
- Compare **approval rates by group**.
- **Disparate impact ratio** = group A rate ÷ group B rate; below **0.8** is a warning sign.
- Add **confidence intervals / significance tests** to check the gap is not due to chance.
