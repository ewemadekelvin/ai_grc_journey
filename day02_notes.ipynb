# Day 2–3 Notes – Google AI Essentials, Module 2

## 1. Where AI saves time
- Summarising a large backlog of emails after time away from work.
- **Don't rely on it blindly:** check summaries for missed deadlines or requests.
- **Privacy risk:** don't paste confidential work data into public AI tools —
  follow the company's AI acceptable-use policy.

## 2. Human-in-the-loop
A human reviews and approves AI output before it is acted on.
- **Example:** a shop's AI sees hoodie sales drop in summer and suggests
  dropping the product. A human knows the dip is **seasonal**.
- **Hallucination** = AI confidently states something false or made up.
- **Missing context** = AI reads real data but draws the wrong conclusion
  (like a regression without seasonal dummies).

## 3. Garbage in, garbage out
The quality of an AI's output depends on the quality of its training data.
Incomplete, outdated or biased data → unreliable or biased output.
