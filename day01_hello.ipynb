# Day 1 Notes – AI GRC Journey

## 1. What surprised me
Artificial intelligence: a machine using maths and logic to perform cognitive tasks.
AI is still at its early stage, and its growth rate is comparable to that of the internet.

## 2. What is AI? (my definition)
AI is a computer system that **learns patterns from data**, so it can handle complex
tasks such as predicting, recognising or generating content that normally need
human intelligence.

- AI started in the **1950s**
- **Generative AI** reached the public in **2022** (e.g. ChatGPT, Claude)

## 3. How AI is governed – key frameworks
1. **NIST AI RMF** – US framework with four functions: *Govern, Map, Measure, Manage*
2. **EU AI Act** – a law that sorts AI systems by **risk level**
3. **ISO/IEC 42001** – a certifiable **management system standard** for AI

## Key points on managing AI risk
- **NB:** A model is **trained, tested, then frozen** before release.
  If it is **retrained** later, the change is **documented**.
- **Model drift:** when a frozen model's accuracy drops over time because
  **the real world (and its data) changes** while the model stays the same.
