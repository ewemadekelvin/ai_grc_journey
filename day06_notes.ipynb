- **The AI divide:** researchers in countries with limited internet, computing power
  or funding may fall further behind — an infrastructure and cost gap.

## 3. My question: can an AI model be "infected" like a virus?
Yes — this is **AI security**:
| Attack | What happens |
|---|---|
| **Data poisoning** | Bad data slipped into training → model learns wrong or hidden behaviour |
| **Prompt injection** | Hidden instructions in a web page, email or document hijack the AI |
| **Malicious model files** | Downloaded models can contain hidden code that runs on my computer |
| **Adversarial examples** | Tiny invisible changes make the model misclassify |
| **Jailbreaking** | Prompts that get around the model's safety rules |

Frameworks to study in Phase 2: **OWASP Top 10 for LLM Applications** (prompt injection = #1 risk)
and **MITRE ATLAS**.
