# Course 2 Notes – Maximize Productivity With AI Tools ✅ (5 Oct 2026)

## 1. AI augments, it doesn't replace
- AI automates tasks and makes work easier.
- **Human oversight** stays essential: to check accuracy, fit the audience,
  and stay within **legal and policy** limits.

## 2. Three ways organisations adopt AI
| Type | What it is | Example |
|---|---|---|
| Stand-alone | Separate app | Gemini / ChatGPT website |
| Integrated | Built into existing software | Gemini in Docs, Copilot in Word |
| Custom | Built for one organisation, on its own data | Internal company chatbot |
- Best practice: **pilot in one team → measure → scale with an AI policy.**

## 3. Generative AI by output
- **Text**, **code**, **image and video**
- Image/video tools mostly use **diffusion models** (turn noise into an image step by step).

## 4. Two separate risks
- **Knowledge cutoff:** model only knows data up to its training date → **outdated** answers.
- **Hallucination:** model **makes things up** confidently — can happen anytime.
- **Reduce both:** grounding (search / own documents) + check important facts.
